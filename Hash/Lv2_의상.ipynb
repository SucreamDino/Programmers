```python
clothes = [["yellow_hat", "headgear"], ["blue_sunglasses", "eyewear"], ["green_turban", "headgear"]]
```

```python
def solution(clothes):
    dict = {}

    for _, type in clothes:
        if type not in dict:
            dict[type] = 2
        else:
            dict[type] += 1
    
    count = 1
    for num in dict.values():
        count *= num
    
    return count-1
```

```python
solution(clothes)
```
