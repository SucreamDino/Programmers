```python
participant = ["leo", "kiki", "eden"]
completion = ["eden", "kiki"]
```

```python
def solution(participant, completion):
    participant.sort() #.sort()는 리스트를 오름차순으로 정렬
    completion.sort()

    for a, b in zip(participant, completion): #zip(a,b)는 a와 b의 요소를 튜플로 묶어서 반환하는 함수
        if a != b:
            return a
    
    return participant[-1]

```

```python
solution(participant, completion)
```
