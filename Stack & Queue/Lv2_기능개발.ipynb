```python
from math import ceil 

def solution(progresses, speeds):
    fin_date = []
    
    for x, y in zip(progresses, speeds):
        fin_date.append(ceil((100-x) / y))
        
        answer = []
        
        current = fin_date[0]
        count = 1
        
        for day in fin_date[1:]:
            if day <= current:
                count += 1
            else:
                answer.append(count)
                current = day
                count = 1
                
    answer.append(count)
        
    return answer
```

```python
progresses = [93, 30, 55]
speeds = [1, 30, 5]

answer = solution(progresses, speeds)
print(answer)
```
